## Data Source

**Provider:** Ministry of Environment (MOENV)  
**Dataset:** AQX_P_432 — Air Quality Index (AQI)  
**Access method:** MOENV Open Data API  
**Access date:** 2026-10-01  
**Temporal resolution:** Hourly  
**Raw snapshot:** `data/raw/moenv_AQX_P_432_2026-10-01_sample.json`

### Initial limitation

This snapshot represents the data available at the retrieval time and is not a complete historical dataset.

```python
from pathlib import Path
from getpass import getpass

import json
import pandas as pd
import requests
```

```python
API_KEY = getpass.getpass("Enter your MOENV API key: ")
```

```python
url = "https://data.moenv.gov.tw/api/v2/aqx_p_432"

params = {
    "limit": 1000,
    "format": "json",
    "api_key": API_KEY,
}

response = requests.get(
    url,
    params=params,
    timeout=30
)

response.raise_for_status()
payload = response.json()

print("HTTP status:", response.status_code)
print("Payload type:", type(payload))

if isinstance(payload, dict):
    print("Top-level keys:", payload.keys())
elif isinstance(payload, list):
    print("Number of returned records:", len(payload))
    print("First record:")
    print(payload[0])
```

```python
if isinstance(payload, dict):
    records = payload.get("records", [])
elif isinstance(payload, list):
    records = payload
else:
    raise TypeError(f"Unexpected payload type: {type(payload)}")

df = pd.DataFrame(records)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(df.head())
```

```python
PROJECT_ROOT = Path.cwd().parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)

RAW_FILE = RAW_DIR / "moenv_AQX_P_432_2026-10-01_sample.json"

print(RAW_FILE)
```

```python
with open(RAW_FILE, "w", encoding="utf-8") as f:
    json.dump(
        payload,
        f,
        ensure_ascii=False,
        indent=2
    )

print("Saved:", RAW_FILE)
```
