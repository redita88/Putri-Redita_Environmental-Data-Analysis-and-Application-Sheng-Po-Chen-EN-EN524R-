```python
from pathlib import Path
from getpass import getpass

import json
import pandas as pd
import requests
import urllib3

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
```

```python
# Assignment 3

## Research Question

How does relative humidity (RH) vary with air temperature (AMB_TEMP) in Taiwan?

## Data Source

Taiwan Ministry of Environment (MOENV) Open Data API

Dataset:
AQX_P_35

## Selected Variables

- AMB_TEMP (Air Temperature, °C)
- RH (Relative Humidity, %)
```

```python
import getpass

API_KEY = getpass.getpass("Enter your MOENV API key: ")
print("API key loaded.")
```

```python
url = "https://data.moenv.gov.tw/api/v2/aqx_p_35"

params = {
    "limit": 1000,
    "format": "json",
    "api_key": API_KEY,
}

response = requests.get(
    url,
    params=params,
    timeout=30,
    verify=False
)

response.raise_for_status()

payload = response.json()

print("HTTP status:", response.status_code)
print("Payload type:", type(payload))
```

```python
temp_df = df[
    df["itemengname"] == "AMB_TEMP"
].copy()

temp_df.head()
rh_df = df[
    df["itemengname"] == "RH"
].copy()

rh_df.head()
```

```python
print("Temperature Shape:", temp_df.shape)

print("\nMissing Values:")
print(temp_df.isna().sum())

print("\nData Types:")
print(temp_df.dtypes)
print("RH Shape:", rh_df.shape)

print("\nMissing Values:")
print(rh_df.isna().sum())

print("\nData Types:")
print(rh_df.dtypes)
```

```python
temp_df["concentration"] = pd.to_numeric(
    temp_df["concentration"],
    errors="coerce"
)

rh_df["concentration"] = pd.to_numeric(
    rh_df["concentration"],
    errors="coerce"
)
```

```python
RAW_DIR = Path("data/raw")
RAW_DIR.mkdir(parents=True, exist_ok=True)

RAW_FILE = RAW_DIR / "aqx_p_35_sample.json"

with open(RAW_FILE, "w", encoding="utf-8") as f:
    json.dump(
        payload,
        f,
        ensure_ascii=False,
        indent=2
    )

print("Saved:", RAW_FILE)
```
